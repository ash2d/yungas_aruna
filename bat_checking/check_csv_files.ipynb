```python
from pathlib import Path
import shutil
import pandas as pd
```

```python

PT_folder = '/Users/ash/phd/Bat_Data'
```

```python

src_root = Path(PT_folder)

meta_count = 0
wav_count = 0

for top_level in (p for p in src_root.iterdir() if p.is_dir()):
    # inside each top-level folder, pick the two that do not begin with '1'
    second_level_candidates = [p for p in top_level.iterdir() if p.is_dir() and not p.name.startswith('1')]
    for second_level in second_level_candidates:
        # go through all third-level folders
        for third_level in (p for p in second_level.iterdir() if p.is_dir()):
            # copy meta.csv from Data
            data_dir = third_level / 'Data'
            meta_file = data_dir / 'meta.csv'
            if meta_file.exists():
                df = pd.read_csv(meta_file)
                meta_count += 1

            # copy all .wav from DATA_filtered
            filtered_dir = data_dir / 'DATA'
            # print(f"Checking {filtered_dir}")
            if filtered_dir.exists():
                # print(f"Processing {filtered_dir}")
                for wav in filtered_dir.glob('*.wav'):
                    # print(wav.name)
                    if wav.name in df['IN FILE'].values:
                        wav_count += 1
                if wav_count == len(df):
                    print(f"All .wav files listed in {meta_file} are present in {filtered_dir}.")
                else:
                    print(f"{len(df)-wav_count} .wav files listed in {meta_file} are missing in {filtered_dir}.")
            wav_count = 0  # reset for next iteration

print(f"found [{meta_count}] meta.csv files, should be {10*2*3}.")
```
