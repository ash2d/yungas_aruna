open harddrive - MS - edited by germanan, farm , habitat types 2,3. 3 points within each. open Data, data filtered, then each recording inside, csv file with labels called meta in data folder.

```python
from pathlib import Path
import shutil

# Set destination root (change this to your target folder)
dest_root = Path('/Users/ash/phd/Bat_Data')

src_root = Path(PT_folder)
dest_root.mkdir(parents=True, exist_ok=True)

meta_count = 0
wav_count = 0

for top_level in (p for p in src_root.iterdir() if p.is_dir()):
    # inside each top-level folder, pick the two that do not begin with '1'
    second_level_candidates = [p for p in top_level.iterdir() if p.is_dir() and not p.name.startswith('1')]
    for second_level in second_level_candidates:
        # go through all third-level folders
        for third_level in (p for p in second_level.iterdir() if p.is_dir()):
            # copy meta.csv from Data
            data_dir = third_level / 'Data'
            meta_file = data_dir / 'meta.csv'
            if meta_file.exists():
                dest_meta = dest_root / meta_file.relative_to(src_root)
                dest_meta.parent.mkdir(parents=True, exist_ok=True)
                shutil.copy2(meta_file, dest_meta)
                meta_count += 1

            # copy all .wav from DATA_filtered
            filtered_dir = data_dir / 'DATA'
            # print(f"Checking {filtered_dir}")
            if filtered_dir.exists():
                # print(f"Processing {filtered_dir}")
                for wav in filtered_dir.glob('*.wav'):
                    dest_wav = dest_root / wav.relative_to(src_root)
                    dest_wav.parent.mkdir(parents=True, exist_ok=True)
                    shutil.copy2(wav, dest_wav)
                    wav_count += 1

print(f"Copied {meta_count} meta.csv files and {wav_count} .wav files to {dest_root}")
```
